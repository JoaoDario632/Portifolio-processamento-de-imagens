```python
img = cv2.imread("Foto Atividade.png")
img
```

```python
imgGRAY = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(imgGRAY, cmap="gray")
plt.title("Imagem em tons de CINZA")
plt.axis("off")
plt.show()
```

```python
def quantizar(imagem, bits):
    niveis = 2 ** bits
    ImgFloart = imagem.astype(np.float32)
    quantizada = np.floor(ImgFloart * (niveis - 1) / 255)
    quantizada = quantizada * (255 / (niveis - 1))
    return quantizada.astype(np.uint8)
```

```python
img1bit = quantizar(imgGRAY, 1)
img2bits = quantizar(imgGRAY, 2)
img4bits = quantizar(imgGRAY, 4)
img8bits = quantizar(imgGRAY, 8)
```

```python
fig, ax = plt.subplots(1, 5, figsize=(18, 5))

ax[0].imshow(imgGRAY, cmap="gray", vmin=0, vmax=255)
ax[0].set_title("Original com Tons de cinza")
ax[1].imshow(img1bit, cmap="gray", vmin=0, vmax=255)
ax[1].set_title("1 bi n2 níveis")
ax[2].imshow(img2bits, cmap="gray", vmin=0, vmax=255)
ax[2].set_title("2 bits 4 níveis")
ax[3].imshow(img4bits, cmap="gray", vmin=0, vmax=255)
ax[3].set_title("4 bits 16 níveis")
ax[4].imshow(img8bits, cmap="gray", vmin=0, vmax=255)
ax[4].set_title("8 bits 256 níveis")
for i in range(5):
    ax[i].axis("off")
plt.tight_layout()
fig.savefig("quantizacao.png", dpi=300, bbox_inches="tight")
plt.show()
```
